# Sentencias `break` and `continue`

En ocaciones, es necesario tener más control sobre estructuras de repetición como `for` y `while`. Por ejemplo si queremos salir de un ciclo de repeticiones, saltar la iteración actual en que nos encontramos

# Sentencia `break` con `while`

```python
contador = 1

while contador <= 31:
    if contador != 13:
        print(f"Dia del mes {contador}")
        contador += 1
    else:
        break       

print(f"El ciclo while se interrumpio al llegar al dia {contador}")
```

# Sentencia `continue` con `while`

```python
contador = 1

while contador <= 31:
    if contador == 13:
        valor_omitido = contador
        contador += 1
        continue
    print(f'Dia del mes: {contador}')
    contador += 1

print(f'El ciclo while omitio el dia {valor_omitido}.')
```

# Sentencia `break` con `for`

```python
dbz = ['Goku','Veggeta','Gohan','Freezer','Trunks']

for personaje in dbz:
    if personaje == "Freezer":
        personaje_omitido = personaje
        break
    print(f'Hola {personaje}')

print()
print(f'Peligro detectado!!!! {personaje_omitido} se encuentra cerca😥😥')
```

# Sentencia `continue` con `for`

```python
dbz = ['Goku','Veggeta','Gohan','Freezer','Trunks']

for personaje in dbz:
     if personaje == "Freezer":
            personaje_omitido = personaje
            continue
     print(f'El personajes es: {personaje}')

print()
print(f'El personaje que imitimos fue {personaje_omitido} 😉 ')
```
