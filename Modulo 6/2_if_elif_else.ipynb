# <u>**Estructuras de control**</u>

# Estructura condicional `if`

Permite ejecutar bloques de código basados en si una condición es verdadera o falsa.

Su sintaxis básica es la siguiente:

`if "condición":`

        Bloque de código que se ejecuta si la condición es verdadera

Su sintaxis compuesta es la siguiente:

`if "condición":`

        Bloque de código que se ejecuta si la condición es verdadera
`else:`

        Bloque de código que se ejecuta si la condición es falsa

* Operadores <u>relacionales</u> que se pueden utilizar con el operador condicional `if`:

    * Mayor `>`
    * Mayor igual `>=`
    * Menor `<`
    * Menor igual `<=`
    * Igual `==` `(*comparación*)
    * Diferente `!=`

* Operadores <u>lógicos</u> que se pueden utilizar con el operador condicional `if`:

    * Y `and`
    * O `or`
    * Negación `not`

# Sintáxis básica del `if`

```python
edad = int(input("Ingresa tu edad: "))
if edad >= 18: # if = si (condicional)
    print(f"La persona tiene {edad} años de edad")
    print("La persona es mayor de edad")
```

# Sintáxis compuesta del `if` con `else`

```python
edad = int(input("Ingresa tu edad: "))
licencia_conducir = True

if edad >= 18 and licencia_conducir == True:
    print(f"La persona tiene {edad} años de edad")
    print("La persona es mayor de edad")
    print("La persona se encuentra habilitadad para conducir")
    print(f'Habilitado para conducir: {True}')
else: # else = si no
    licencia_conducir == False
    print(f'La persona tiene {edad} años de edad')
    print("La persona no es mayor de edad")
    print(f'Habilitado para conducir: {False}')
```

# Sintáxis compuesta del `if` - `elif` - `else`

```python
edad = int(input("Ingresa tu edad: "))
licencia_conducir = True


if edad >= 18 and licencia_conducir:
    print(f"La persona tiene {edad} años de edad")
    print("La persona es mayor de edad")
    print(f'Habilitado para conducir: {licencia_conducir}')
elif edad >= 16 and edad < 18 and licencia_conducir:
    print(f"La persona tiene {edad} años de edad")
    print("La persona aún no es mayor de edad")
    print(f'Habilitado para conducir: {licencia_conducir}')
else: # else = si no
    licencia_conducir = False
    print(f'La persona tiene {edad} años de edad')
    print("La persona no es mayor de edad")
    print(f'Habilitado para conducir: {licencia_conducir}')


```

# `if` ternario

El if ternario en Python es una forma compacta de escribir una estructura condicional `if` en una sola línea. Es útil cuando quieres asignar un valor a una variable dependiendo de una condición.

**Sintaxis**

`"sentencia True" if condición else "sentencia False"`

```python
edad = int(input("Ingresa tu edad: "))
licencia_conducir = True

mensaje = "Eres mayor de edad" if edad >= 18 else "Eres menor de edad"
print(mensaje)
```
