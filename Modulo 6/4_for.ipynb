# <u>**Estructuras de control**</u>

# Estructura de repetición `for`

la estructura de control `for` se utiliza para iterar (recorrer) sobre una secuencia, como una lista, una tupla, un diccionario, un conjunto o incluso una cadena de texto. Es muy útil cuando sabes cuántas veces quieres que se repita un bloque de código.

Su sintaxis básica es la siguiente:

`for "variable" in "secuencia":`

        Bloque de código a ejecutar por cada elemento de la secuencia

# Uso de `for` con `dict`

```python
datos_personales = {
    "nombre" : "Goku",
    "apellido" : "Son",
    "raza" : "Sayayin",
    "planeta" : "Veggita"
}

datos_personales
```

## Iterando las "llaves" y "valores" de un `dict`

```python
for key in datos_personales.keys():
    print(key)
```

```python
for values in datos_personales.values():
    print(values)
```

```python
for key, values in datos_personales.items():
    print(f'{key} -- {values}')
```

# Uso de `for` con `range()`

Sintaxis

`range(inicio, fin[no inclusivo], paso[opcional])`

```python
for i in range(4,10):
    print(i)
```

```python
for elemento in range(0, 22, 2):
    print(elemento)
```

```python
for i in range(5):
    print(f"Valor actual en el rango: {i}. Estoy aprendiendo a utilizar la estructura de repetición for en python")
```

# Uso de `for` con `enumerate()`

Sintaxis

`enumerate(secuencia, inicio=0)`

```python
# enumerate lo que hace es asignar un indice y luego un valor 

lenguajes = ["Python", "JavaScript", "C++", "Java"]

for indice, lenguaje in enumerate(lenguajes, 1):
    print(f'{indice}: {lenguaje}')
```

```python
temp_cpu = int(input('Ingrese temperatura: '))

if temp_cpu < 50:
    print('Temperatura Excelente (En reposo)')
elif temp_cpu <= 50:
    print('Temperatura Normal (En carga)')
elif temp_cpu <= 85:
    print('Temperatura Alta (Revisar flujo de aire)')
else:
    print('ALERTA CRÍTICA: Peligro de Thermal Throttling')

estado = "Seguro" if temp_cpu <= 75 else "Riesgo"
print(f'Estado del sistema: {estado} {temp_cpu}°C')


```

```python
servidores = [
    {"nombre": "Srv-Web", "ram_pct": 92, "es_critico": False},
    {"nombre": "Srv-BD-Principal", "ram_pct": 88, "es_critico": True},
    {"nombre": "Srv-Backup", "ram_pct": 45, "es_critico": False},
    {"nombre": "Srv-BD-Secundario", "ram_pct": 78, "es_critico": True},
    {"nombre": "Srv-Logs", "ram_pct": 95, "es_critico": True}
]
```

```python
for servidor in servidores:
    if servidor["ram_pct"] > 85 and servidor["es_critico"] == True:
        print(f'Alerta maxima: {servidor["nombre"]} esta al ({servidor["ram_pct"]})% de RAM y es critico')
    elif servidor["ram_pct"] > 80:
        print(f'Advertencia: {servidor["nombre"]} tiene un uso elevado ({servidor["ram_pct"]})%')
    else:
        print(f'Servidor Nombre {servidor["nombre"]} esta operand estable ({servidor["ram_pct"]})%')
```

```python
tickets = [
    {"id": 101, "area": "Redes", "horas_abierto": 30, "urgente": False},
    {"id": 102, "area": "Seguridad", "horas_abierto": 5, "urgente": False},
    {"id": 103, "area": "Sistemas", "horas_abierto": 10, "urgente": True},
    {"id": 104, "area": "Impresoras", "horas_abierto": 8, "urgente": False}
]
```
