# `list comprehension` y el método `join()`

## `list comprehension`

Un `list comprehension` (comprensión de listas) es una forma concisa y elegante de crear listas en Python. Permite generar una nueva lista aplicando una expresión a cada elemento de una secuencia o iterable (como una lista, un rango, etc.), todo en una sola línea de código.

`Sintaxis básica`



`Sintaxis con condicional`



```python
lista_numeros = [i for i in range(1,11)]

print(lista_numeros)
```

```python
numeros_cuadrados = [elemento**2 for elemento in lista_numeros]

print(numeros_cuadrados)
```

```python
print(lista_numeros)
```

```python
numeros_pares = [numero for numero in lista_numeros if numero % 2 == 0]

print(numeros_pares)
```

## Método `join()`

El método `join()` se utiliza para unir una secuencia de elementos (generalmente una lista de cadenas de texto) en una sola cadena, con un **delimitador** específico entre esos elementos.

`Sintaxis`



```python
personajes = ["goku","tanjiro","lufy","naruto"]
```

```python
print(f'Mis personajes favoritos son: {" - ".join(personajes)}')
```

```python
personajes = [personaje.title() for personaje in personajes]

print(personajes)
```

```python
personajes = [personaje.title() for personaje in personajes]

print(f'Mis personajes favoritos son: {", ".join(personajes)}')
```
