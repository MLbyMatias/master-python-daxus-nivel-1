# 🛒 Sistema de control de inventario 

### 🏗️ Funcionalidades

✅ Menú interactivo para que el usuario elija acciones

✅ Registro de múltiples productos sin un límite predefinido

✅ Eliminación de productos del inventario

✅ Visualización del inventario actualizado

✅ Uso de bucles (`while`, `for`) para recorrer los productos

✅ Uso de condicionales (`if`, `elif`, `else`) para gestionar las opciones del menú


* notebook.output - scrolling

* line numbres - notebook

```python
import os


# Definir variables globales
inventario = []
IVA = 0.19

# Utilizo un while True para mantener la condicion "viva"
while True:
    print("\n🛒 Bienvenidos")
    print("1. Agregar Producto")
    print("2. Ver Inventario")
    print("3. Eliminar Producto")
    print("4. Salir")

    opcion = input("Elija una opción (1-4)").strip() 

    
    #Freno de mano, esperando la respuesta de entrada del usuario
    
    #En consola se verá limpio: Elija una opción (1-4): 1

    #Si el usuario escribe en la entrada del input opcion "1" lo dirige aca

    
    if opcion == "1":
        # Solicitando información al usuario
        producto = input("Ingrese el nombre del producto: ").title().strip()
        cantidad= int(input("Ingrese la cantidad: "))
        precio_unitario = float(input("Precio del producto: "))


        # Calculo del valor total del inventario
        valor_total = cantidad * precio_unitario

        # Almacenamos nuestro producto en una estructura de datos que seria un diccionario 
        producto = {
            "nombre":producto,
            "cantidad":cantidad,
            "precio":precio_unitario,
            "valor_total":valor_total
        }

        # Agregamos nuestro producto a nuestra lista con el metodo append()
        # append de por si lo agrega como ultimo elemento, pero como esta vacia la lista lo agrega como primero
        # El inventario ahora contendria el primero producto de diccionario que se agregue
        inventario.append(producto)
        print(f'Producto: {producto["nombre"]} agregado correctamente')

    # Si el usuario escribe en la entrada del input opcion "2" lo dirige aca
    elif opcion == "2":
        if not inventario:  # Como inventario esta vacio esto equivale a False, ante poner operador "not" se transforma en True
                            # Estariamos confirmando con un True de que la lista inventario esta vacia
            print("El inventario esta vacio")            
        else:
            subtotal = 0 # Aca iremos sumando el "valor_total" de cada producto
            print("\n Productos en Inventario")
            
            # n_producto le asigna una posicion al diccionario con el producto
            # prod contiene el diccionario completo el producto
            for n_producto, prod in enumerate(inventario, 1):
                print(f'#{n_producto} {prod["nombre"]} | {prod["cantidad"]} | ${prod["precio"]} | {prod["valor_total"]:.2f}') 
                 # Recorre la lista y le asigna un numero a cada elemento
                subtotal += prod["valor_total"] # Incrementamos la variable subtotal con lo que contiene "valor_total" del producto

            total_iva = subtotal * IVA # Calculamos el iva de cada producto que exista
            total_final = subtotal + total_iva # Le agregamos el IVA al producto


            # Mostrar Inventario con productos
            print("-" * 40)
            print(f'El valor total es de: ${subtotal:.2f}')
            print(f'El IVA total de esta boleta es de: {total_iva:.2f}')
            print(f'El valor con IVA total es de: {total_final:.2f}')

    # Si el usuario escribe en la entrada del input opcion "3" lo dirige aca
    elif opcion == "3":
        if not inventario:
            # Como inventario esta vacio esto equivale a False, ante poner operador "not" se transforma en True
            # Estariamos confirmando con un True de que la lista inventario esta vacia
            print("No hay productos para eliminar")
        else:
            producto_eliminar = input("Nombre del producto a eliminar").title().strip()
            encontrado = False # Esto es como un interruptor, lo dejamos apagado
                               # Por si colocamos un print("No enconctrado") dentro del for
                               # Imprimiria esto por cada producto no coincida mientras lo busca

            for prod in inventario: # Revisamos cada prod(producto) de la lista
                if prod["nombre"] == producto_eliminar: # Comparamos el valor "nombre" de ese producto a eliminar
                    inventario.remove(prod) # remove() elimina la ocurrencia selañada que este en la lista y lo borra
                    print(f'Producto {producto_eliminar} eliminado')
                    encontrado = True # Encendemos nuestro interruptor, (confirmamos que si existia el producto)
                    break # Detenemos y aplicamos freno de mano ya que hace falta seguir revisando la lista

                if not encontrado:
                    print("Producto no encontrado") # Se ejecuta cuando el bucle termina

    elif opcion == "4":
        print("\n!Hasta luego! Nos vemos Pronto...")
        break # Romple el ciclo del while principal que es de las opciones
    else:
        print("\n Opción no valida. elije una opcion de 1 a 4 (1-4).")  

                
```
