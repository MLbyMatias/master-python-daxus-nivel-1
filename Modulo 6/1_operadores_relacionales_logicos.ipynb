# Operadores relacionales

* Mayor `>`
* Mayor igual `>=`
* Menor `<`
* Menor igual `<=`
* Igual `==` (*comparación*)
* Diferente `!=`

```python
edad_padre = 50
edad_hija = 12

print(edad_hija > edad_padre)
```

```python
distancia_1 = 3.5
distancia_2 = 8

print(distancia_2 >= distancia_1)
```

```python
ciudad_usuario_1 = "Bogota"
ciudad_usuario_2 = "Cali"

print(ciudad_usuario_1 == ciudad_usuario_2)
```

```python
temperatura_1 = 16
temperatura_2 = 18

print(temperatura_1 != temperatura_2)
```

## `and`

## `or`
