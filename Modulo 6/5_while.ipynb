# Estructura de repetición `while`

El while es una estructura de control en programación que ejecuta un bloque de código de manera repetida mientras una condición sea verdadera.

<u>Sintaxis</u>

`while "condición":`

        Bloque de código a ejecutar mientras la condición sea verdadera

<u>Diferencia entre `while` y `for`</u>

La condición de parada para el `while` es explícita. Se repite mientras la condición sea verdadera. No depende de un rango específico o de una secuencia.

# `while` con un contador

```python
contador = 0

while contador < 5:
    print("Hola mundo python")
    print(f'Contador al llegar al ciclo while: {contador}')
    # contador = contador + 1
    contador += 1
    print(f'Contador al salir del ciclo while: {contador}\n')

print("Fin de la ejecucion del while")


    
```

```python
vida_personaje = 7

while vida_personaje > 0:
    print(f'Vida actual del personaje: {vida_personaje}')
    vida_personaje -= 1 
    print(f'Recibi un ataque. Pierdo un punto de vida')
    print(f'Puntos de vida restantes: {vida_personaje}')
    print("--------------------------------------------------")

print("Perdiste la partida 😞")
```
