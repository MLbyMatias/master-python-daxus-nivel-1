# Tipado dinámico

El tipado dinámico es una característica de algunos lenguajes de programación, que no exige la declaración del tipo de dato, dado que éstos son capaces de inferir dinámicamente el tipo de dato a utilizar, pudiendo cambiarlo durante la compilación o la ejecución del programa.

`Python` es un lenguaje de tipado `dinámico` y `débil`, es decir, no requiere definir el tipo de dato de una variable al momento de declararla.

```python
nombre_curso = "Master Python"

print(nombre_curso)
print(type(nombre_curso))
```

```python
nombre_curso = True

print(nombre_curso)
print(type(nombre_curso))
```

```python
nombre_curso = 77

print(nombre_curso)
print(type(nombre_curso))
```
