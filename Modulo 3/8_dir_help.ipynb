# Funciones `dir()` y `help()`

Las funciones `dir()` y `help()` en Python son herramientas muy útiles para explorar objetos y obtener información sobre ellos, especialmente cuando estás trabajando en un entorno interactivo (como un intérprete o un cuaderno de Jupyter).

## Función `dir()`

En Python se utiliza para obtener una lista de los atributos y métodos (funciones) disponibles para un objeto.

```python
print(dir(input))
```

## Función `help()`

se usa para obtener información más detallada sobre un objeto, módulo, función, o clase. Muestra la documentación asociada al objeto, lo que incluye detalles sobre cómo se usa, qué parámetros acepta, qué devuelve, etc.

```python
print(help(print))
```
