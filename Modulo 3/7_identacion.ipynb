# La `indentación`

En Python, la indentación no es solo una cuestión de estilo, sino que tiene una función semántica importante. A diferencia de otros lenguajes de programación, que utilizan corchetes `{}` o palabras clave para delimitar bloques, Python utiliza la indentación para marcar el principio y el final bloques de código.
