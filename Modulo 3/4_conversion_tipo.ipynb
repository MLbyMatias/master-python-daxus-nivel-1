# Conversión de tipos ó `casting`

El `casting` se refiere al proceso de cambiar un valor de un tipo de dato a otro.

# `casting` a `int`

El tipo de dato int representa números enteros. Puedes convertir otros tipos a int si es posible (por ejemplo, cadenas numéricas o valores flotantes).

```python
numero_float = 3.14

print(int(numero_float))
```

```python
numero_str = input("Ingrese un número: ")

print(int(numero_str))
```

```python
booleano = False
print(int(booleano))
```

```python
string = "Matias"

print(int(string))
```

# `casting` a `float`

El tipo de dato float representa números decimales. Puedes convertir otros tipos a float si es posible.

```python
numero_entero = 5

print(float(numero_entero))
```

```python
num_str = "3.1416"

print(float(num_str))
```

```python
booleano = True

print(float(booleano))
```

```python
string = "Python"

print(float(string))
```

# `casting` a `bool`

El tipo de dato bool representa valores de verdadero o falso. Puedes convertir otros tipos a bool. False se considera 0 y True se considera cualquier valor diferente de 0 o vacío.

```python
num_int = 0

print(bool(num_int))
```

```python
numero_float = 1.0

print(bool(numero_float))
```

```python
string_vacio = ""

print(bool(string_vacio))
```

```python
string_vacio = "Holaaa"

print(bool(string_vacio))
```

# `casting` a `str`

El tipo de dato str representa cadenas de texto. Puedes convertir otros tipos a str fácilmente.

```python
num_int = 123

print(type(str(num_int)))
```

```python
num_float = 2.26834

print(type(str(num_float)))
```

```python
booleano = True

print(type(str(booleano)))
```
