```python
mascotas = []
ancho = 1256

from fpdf import FPDF
from PyPDF2 import PdfReader, PdfWriter
import pytz

def obtener_edad():
    while True:
        try:
            edad = int(input("Ingrese la edad de la mascota: "))
            if 0 <= edad <= 21:
                return edad
            print(f"Edad fuera de rango, edad ingresada {edad} es invalida.")
        except ValueError:
            print("Por favor, ingrese un número entero válido para la edad.")

def obtener_peso():
    while True:
        try:
            peso_mascota = float(input("Ingrese el peso de la mascota (KG): "))
            if 0 <= peso_mascota:
                return peso_mascota
            print(f'Ingrese un peso valido. Peso ingresado {peso_mascota}')
        except ValueError:
            print("Por favor, Ingrese un numero decimal valido para el peso.")


def agregar_mascota(lista_mascotas):
    while True:
        nombre = input("Ingrese nombre de la mascota: ").title().strip()
        if nombre:
            break
        print("El nombre de la mascota no puede estar vacío. Por favor, ingrese un nombre válido.")

    if any(mascota["nombre"].lower() == nombre.lower() for mascota in lista_mascotas):
        print("Ya existe una mascota con ese nombre.")
        return

    while True:
        genero = input("Ingrese el género de la mascota.(M) = Macho, (H) = Hembra").title().strip()
        if genero == "M":
            genero = "Macho"
            break
        elif genero == "H":
            genero = "Hembra"
            break
        else:
            print("Opción inválida. Ingrese solo M o H para el género.")

    while True:
        especie = input("Ingrese la especie del animal (Gato, Perro, Etc): ").title().strip()
        if especie:
            break
        print("La especie de animal no puede estar vacío. Por favor, ingrese un tipo válido.")

    while True:
        raza = input("Ingrese nombre de raza: ").title().strip()
        if raza:
            break
        print("El nombre de la raza no puede estar vacío. Por favor, ingrese un nombre válido.")

    edad = obtener_edad()

    peso = obtener_peso()

    while True:
        color_pelo = input("Ingrese color de pelo del animal: ").title().strip()
        if color_pelo:
            break
        print("El color de pelo no puede estar vacio.")

    while True:
        tipo_pelo = input("Ingresa el tipo de pelo del animal: ").title().strip()
        if tipo_pelo:
            break
        print("El tipo de pelo no puede estar vacio.")

    nueva_mascota = {
        "nombre": nombre,
        "genero":genero,
        "especie": especie,
        "raza": raza,
        "edad": edad,
        "peso": peso,
        "color_pelo": color_pelo,
        "tipo_pelo": tipo_pelo
    }

    lista_mascotas.append(nueva_mascota)

    print(f"\n¡{nombre} registrado con éxito!")


def listado_mascotas(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas ingresadas aún")
        return

   
    header = (
        f"| {'#':^3} "
        f"| {'NOMBRE':<18} "
        f"| {'GENERO':<10} "
        f"| {'ESPECIE':<12} "
        f"| {'RAZA':<15} "
        f"| {'EDAD':^6} "
        f"| {'PESO':^6}"
        f"| {'COLOR':<15} "
        f"| {'PELAJE':<15} |"
    )
    
    ancho = len(header)

    print("\n" + "=" * ancho)
    print("--- LISTADO DE MASCOTAS ---".center(ancho))
    print("=" * ancho)
    print(header)
    print("-" * ancho)

    for numero, mascota in enumerate(lista_mascotas, 1):
        print(
            f"| {numero:^3} "
            f"| {mascota['nombre'][:18]:<18} "
            f"| {mascota['genero'][:10]:<10} "
            f"| {mascota['especie'][:12]:<12} "
            f"| {mascota['raza'][:15]:<15} "
            f"| {mascota['edad']:^6} "
            f"| {mascota['peso']:^6}"
            f"| {mascota['color_pelo'][:15]:<15} "
            f"| {mascota['tipo_pelo'][:15]:<15} |"
        )

    print("=" * ancho)
    print(f"Total de mascotas registradas: {len(lista_mascotas)}")


def editar_mascota(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas")
    else:
        mascota_editar = input("Ingrese nombre de la mascota: ").lower().strip()
        mascota_encontrada = list(filter(lambda m: m["nombre"].lower() == mascota_editar, lista_mascotas))

        for mascota in mascota_encontrada:
                print(f"Mascota con nombre: {mascota_editar} encontrada")
                print("Ingrese los nuevos datos de la mascota:")

                while True:
                    genero = input("Ingrese el género de la mascota.(M) = Macho, (H) = Hembra").title().strip()
                    if genero == "M":
                        mascota["genero"] = "Macho"
                        break
                    elif genero == "H":
                        mascota["genero"] == "Hembra"
                    else:
                        print("Opción inválida. Ingrese solo M o H para el género.")

                while True:
                    especie = input("Ingrese la especie del animal (Gato, Perro, Etc): ").title().strip()
                    if especie:
                        mascota["especie"] = especie
                        break
                    print("La especie de animal no puede estar vacío. Por favor, ingrese una especie válida.")

                while True:
                    raza = input("Ingrese nombre de raza: ").title().strip()
                    if raza:
                        mascota["raza"] = raza
                        break
                    print("El nombre de la raza no puede estar vacío. Por favor, ingrese un nombre válido.")

                mascota["edad"] = obtener_edad()

                mascota["peso"] = obtener_peso()

                while True:
                    color_pelo = input("Ingrese color de pelo del animal: ").title().strip()
                    if color_pelo:
                        mascota["color_pelo"] = color_pelo
                        break
                    print("El color de pelo no puede estar vacio.")
                
                while True:
                    tipo_pelo = input("Ingresa el tipo de pelo del animal: ").title().strip()
                    if tipo_pelo:
                        mascota["tipo_pelo"] = tipo_pelo
                        break
                    print("El tipo de pelo no puede estar vacio.")

                print(f"\n¡{mascota_editar} actualizado con éxito!")
                mascota_encontrada = True
                break

        if not mascota_encontrada:
            print("Mascota no encontrada")


def eliminar_mascota(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas")
    else:
        mascota_eliminar = input("Ingrese nombre de la mascota: ").lower().strip()
        mascota_encontrada = list(filter(lambda m: m["nombre"].lower() == mascota_eliminar, lista_mascotas))

        for mascota in mascota_encontrada:
            lista_mascotas.remove(mascota)
            print(f"Mascota con nombre: {mascota_eliminar} eliminada de la lista")
            mascota_encontrada = True
            break

        if not mascota_encontrada:
            print("Mascota no encontrada")


def limpiar_listado(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas")
        return

    confirmacion = input("¿Desea eliminar por completo la lista actual?. Ingrese 's' para la eliminación ").lower().strip()
    if confirmacion == "s":
        lista_mascotas.clear()
        print("Lista de mascotas eliminada correctamente")
    else:
        print("Operación cancelada")

def eliminar_ultimo(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas")
        return

    borrar_ultimo = input("¿Desea borrar el ultimo elemento?. Ingrese 's' para la eliminación de este ultimo").lower()
    if borrar_ultimo == "s":
        mascota_eliminada = lista_mascotas.pop()
        print(f'Mascota eliminada con exito: {mascota_eliminada["nombre"]}')
    else:
        print("Operación cancelada")


def buscar_mascota(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas.")
        return

    nombre_buscado = input("Ingrese nombre de la mascota: ").lower().strip()

    encontrado = list(filter(lambda m: m["nombre"].lower().strip() == nombre_buscado, lista_mascotas))
    if encontrado:
        mascota = encontrado[0]
        print(f"\nMascota encontrada: {mascota['nombre']}")
        print(f"Genero:{mascota['genero']}")
        print(f"Especie: {mascota['especie']}")
        print(f"Raza: {mascota['raza']}")
        print(f"Edad: {mascota['edad']} años")
        print(f"Peso: {mascota['peso']}")
        print(f"Color de pelaje: {mascota['color_pelo']}")
        print(f"Tipo de pelaje: {mascota['tipo_pelo']}")
        print("=" * 30)
        return

    print(f"La mascota con nombre {nombre_buscado} no está en el registro")
    

def generar_pdf(lista_mascotas):
    if not lista_mascotas:
        print("No hay mascotas registradas para generar el PDF.")
        return

    from datetime import datetime


    zona = list(filter(lambda x: "Santiago" in x,pytz.all_timezones))[0]

    zona_horaria = pytz.timezone(zona)

    fecha_actual = datetime.now(zona_horaria).strftime("%d/%m/%Y %H:%M:%S")

    pdf = FPDF()
    pdf.add_page()


    pdf.set_font("Arial", "B", 16)
    pdf.cell(200,10, txt="Ficha de Mascota", ln=True, align="C")
    pdf.ln(10)


    for numero, mascota in enumerate(lista_mascotas, 1):
        pdf.set_font("Arial", "B", 12)
        pdf.cell(200, 8, txt=f"# {numero} - Nombre: {mascota['nombre']}", ln=True, align="L")

        pdf.set_font("Arial", "", 10)

        pdf.set_font("Arial", "B", 10)
        pdf.cell(25, 6, txt=" * Genero:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['genero']}", ln=True, align="L")

        
        pdf.set_font("Arial", "B", 10)
        pdf.cell(25, 6, txt=" * Especie:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['especie']}", ln=True, align="L")

       
        pdf.set_font("Arial", "B", 10)
        pdf.cell(25, 6, txt=" * Raza:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['raza']}", ln=True, align="L")

      
        pdf.set_font("Arial", "B", 10)
        pdf.cell(25, 6, txt=" * Edad:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['edad']} años", ln=True, align="L")

        pdf.set_font("Arial", "B", 10)
        pdf.cell(25, 6, txt=" * Peso (KG):", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['peso']}", ln=True, align="L")

        
        pdf.set_font("Arial", "B", 10)
        pdf.cell(32, 6, txt=" * Color de pelaje:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['color_pelo']}", ln=True, align="L")

        pdf.set_font("Arial", "B", 10)
        pdf.cell(32, 6, txt=" * Tipo de pelaje:", ln=False, align="L")
        pdf.set_font("Arial", "", 10)
        pdf.cell(200, 6, txt=f"{mascota['tipo_pelo']}", ln=True, align="L")

    pdf.set_y(270)
    pdf.set_font("Arial", "", 10)
    pdf.cell(200, 6, txt=f"Generado el: {fecha_actual}", ln=True, align="R")
    pdf.ln(5)


    pdf.output("listado_mascotas.pdf")
    print("PDF generado con éxito.")

while True:
    print("\n" + "=" * 30)
    print("SISTEMA DE CONTROL DE MASCOTAS")
    print("=" * 30)
    print("1 - Agregar Mascota")
    print("2 - Ver Listado de Mascotas")
    print("3 - Eliminar Mascota")
    print("4 - Editar Mascota")
    print("5 - Buscar Mascota")
    print("6 - Generar PDF")
    print("7 - Limpiar Listado")
    print("8 - Eliminar Ultima Mascota")
    print("9 - Salir")

    opcion = input("\nSelecciona una opción (1-9): ").strip()

    if opcion == "1":
        agregar_mascota(mascotas)
    elif opcion == "2":
        listado_mascotas(mascotas)
    elif opcion == "3":
        eliminar_mascota(mascotas)
    elif opcion == "4":
        editar_mascota(mascotas)
    elif opcion == "5":
        buscar_mascota(mascotas)
    elif opcion == "6":
        generar_pdf(mascotas)
    elif opcion == "7":
        limpiar_listado(mascotas)
    elif opcion == "8":
        eliminar_ultimo(mascotas)
    elif opcion == "9":
        print("Hasta pronto...")
        break
    else:
        print("Opción no válida. Ingrese una opción del 1 al 9.")
```
