# `Funciones`

`Bloques` de código ***reutilizable***.

* `Built-in`: son funciones que vienen predefinidas en Python. No necesitas definirlas, ya que están disponibles por defecto.

* `Custom` / `personalizadas`: son funciones que tú mismo defines usando la palabra clave `def`. Pueden recibir o no parámetros y devolver un valor o no (`None`) un valor.


**Sintaxis**



## Tipos de parámetros

* ***Posicionales***

* ***Con valor predeterminado***

* ***Nombrados***

### Parámetros `posicionales`

Este tipo de parámetros son los que se pasan a la función en el <u>mismo orden</u> en que fueron declarados.

```python
def saludo_usuario(nombre,apellido):
    print(f'{nombre.title()} {apellido.title()}, Bienvendo al curso de Python!!!')
    print(f'{nombre.title()}, Ya te encuentras registrado en el sistema 😏😏')

saludo_usuario("Matias","Aguayo")
```

### Parámetros con `valor predeterminado`

Son parámetros que tienen un valor predeterminado. Si no se pasan al llamar a la función, se usa el valor predeterminado.

```python
def registro_empleado(edad, nombre, estado_civil = "Soltero", compania = "Python Corp.", status = "Activo"):
    print(f'El empleado {nombre} tiene la edad {edad} años de edad, y su estado civil es {estado_civil}, Ingresa a la compañia {compania} y su estado es {status}')
```

```python
# Invocar función
registro_empleado(32,"Pedro")
registro_empleado(18,nombre="juan",estado_civil="divorciado",compania="LG",status="activo")
# registro_empleado(nombre="juan",18,estado_civil="divorciado",compania="LG",status="activo")
# registro_empleado(25,"Matias",estado_civil="Soltero",status="Capacitacion")
```

### Parámetros `nombrados`

Son los parámetros que se pasan <u>explícitamente</u> por su nombre <u>al llamar a la función</u>, sin importar el orden.

```python
def saludo_usuario(nombre,apellido):
    print(f'{nombre.title()} {apellido.title()}, Bienvendo al curso de Python!!!')
    print(f'{nombre.title()}, Ya te encuentras registrado en el sistema 😏😏')

saludo_usuario(apellido="Aguayo",nombre="Matias")
```

```python
def saludo_usuario():

    nombre = input("Ingresa tu nombre")
    apellido = input("Ingresa tu apellido")
    edad = int(input("Ingresa tu edad"))

    print(f'{nombre.title()} {apellido.title()} {edad} , Bienvendo al curso de Python!!!')
    print(f'{nombre.title()} {apellido.title()}, Ya te encuentras registrado en el sistema 😏😏')
```

```python
saludo_usuario()
```
