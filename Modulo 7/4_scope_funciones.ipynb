## `Local Scope`

El alcance local se refiere a las variables que están definidas <u>dentro de una función</u>. Estas variables solo <u>son</u> accesibles dentro de la función.

```python
def function():
     x = 42
     print(f'Valor de la variable definido dentro de la funcion {x}')
     # La variable x tiene un scope local o una alcance no se puede llamar afuera esa variable es decir es solo accesible dentro de la funcion
         
function()
# print(x)
```

```python
function()
```

```python
def validar_palabra():
    while True:

        palabra = input("Ingresa una palabra: ").title()

        if palabra == "Python":
            print(f'Excelente! escribiste "{palabra}" la palabra representa el lenguaje de programacion más utilizado')
            break
        else:
            print(f'Escribiste "{palabra}. No es la palabra que estabamos esperando"')

    return True

validar_palabra()
# print(palabra)
    
```

## `Global Scope`

Variables que están definidas <u>fuera</u> de cualquier función. Son accesibles desde cualquier parte del código, incluso dentro de las funciones, a menos que haya una variable local con el mismo nombre.

```python
variable_global = 42

def mi_funcion():
    print(f'Accediendo desde la funcion "mi funcion a la variable global:"{variable_global}')

def otra_funcion():
    print(f'Accediendo desde la funcion "otra_funcion a la variable global:"{variable_global}')

mi_funcion()
otra_funcion()
print(variable_global)
```

```python
variable_global
```

## `Scope de las funciones anidadas (Enclosing Scope)`

Si una función está definida dentro de otra, las variables de la función exterior son accesibles desde la interior, <u>pero no al revés</u>.

```python
def funcion_externa():
    variable_externa  = 10
    print(f'Llamaa desde la funcion externa') 
    print(f'Variable externa: {variable_externa}')
    print(var_interna)
    print()

    def funcion_interna():
        var_interna = 4
        print(f'Llamaa desde la funcion interna') 
        print(f'Variable interna: {var_interna}')
        print(f'Variable externa: {variable_externa}')

    funcion_interna()
```

```python
funcion_externa()
```

## `global`

Permite modificar una <u>variable global</u> dentro de una función.

```python
variable_global = 5

def modificar_global():
    global variable_global
    variable_global  = 10

modificar_global()
print(variable_global)
```

## `nonglobal`

Se utiliza cuando se quiere <u>modificar una variable de una función exterior</u> (pero no global) desde una función interna.

```python
def funcion_exterior():
    variable_externa = 20
    print(f'Llamaa desde la funcion exterior') 
    print(f'Variable exterior: {variable_externa}')

    def funcion_interna():
        nonlocal variable_externa
        variable_externa = 25
        print(f'Llamaa desde la funcion exterior') 
        print(f'Variable exterior: {variable_externa}')

    funcion_interna()
    
    
```

```python
funcion_exterior()
```
