# `*args` - Parámetros posicionales arbitrarios

`*args` permite que una función reciba un número variable de argumentos posicionales (es decir, sin nombrar). Cuando usas `*args`, todos los argumentos que se pasen a la función (además de los parámetros definidos antes de *args) se agruparán en una ***tupla***.

```python
def suma_dos(num1,num2):
    respuesta = num1 + num2
    return respuesta

print(suma_dos(5,3))
# print(suma_dos(5,3,3))
```

```python
def suma_usuario(nombre, *args):
    print(f'{nombre.title()}, la suma de los numeros {args} es: {sum(args)}')

# Invocar la funcion

suma_usuario("matias",1,2,3,4)
suma_usuario("felipe",100,200,300,400)
```

```python
def sumar_numeros(*args):
    total = 0
    for numero in args:
        total += numero
    return total

sumar_numeros(30,40,50,60)
```

# `**kwargs` - Parámetros posicionales arbitrarios

`**kwargs` funciona de manera similar a ***args**, pero en lugar de recibir argumentos posicionales, recibe argumentos con nombre (pares clave-valor). Estos se agrupan en un ***diccionario***, donde las claves son los nombres de los parámetros y los valores son los valores pasados.

```python
def mostrar_datos_usuario(**kwargs):
    for k, v in kwargs.items():
        print(f'{k.title()} : {v.title()}')
```

```python
mostrar_datos_usuario(nombre = "Matias", apellido = "Aguayo", ciudad = "Santiago",pais = "chile")
```



#### **1.** Parámetros ***posicionales***.

#### **2.** Parámetros con ***valores por defecto***.

#### **3.** ****args***

#### **4.** Parámetros ***nombrados***.

#### **5.** *****kwargs***

# Ejemplo con todos los tipos de `parámetros`

```python
def mi_funcion(posicional, valor_por_defecto = 2, *args, nombrado, por_defecto = 4, **kwargs):
    print(posicional,valor_por_defecto,args,nombrado,por_defecto,kwargs)
```

```python
# Invocar la funcion
mi_funcion(1,3,5,6, nombrado= 7 , nombre = "Matias",)
```

`posicional`: parámetro posicional obligatorio.

`valor_por_defecto`: parámetro posicional con valor por defecto.

`*args`: recoge cualquier número adicional de parámetros posicionales.

`nombrado`: parámetro obligatorio que solo se puede pasar por nombre (keyword).

`por_defecto_con_nombre`: parámetro con valor por defecto, que solo puede ser pasado por nombre.

`**kwargs`: recoge cualquier número de argumentos con nombre como un diccionario.

```python
def mostrar_tecnologias(*args):
    for tecnologia in args:
        print(f'{tecnologia}')
        pass

mostrar_tecnologias("Python", "SQL", "Git", "Docker")
```

```python
def mostrar_ficha(**kwargs):
    for clave, valor in kwargs.items():
        print(f'{clave.title()} : {valor.upper()}')


mostrar_ficha(procesador = "ryzen 7 5700X", RAM = "32GB", GPU = "rtx 5060")
```
