# `Funciones`

`Bloques` de código ***reutilizable***.

* `Built-in`: son funciones que vienen predefinidas en Python. No necesitas definirlas, ya que están disponibles por defecto.

* `Custom` / `personalizadas`: son funciones que tú mismo defines usando la palabra clave `def`. Pueden recibir o no parámetros y devolver un valor o no (`None`) un valor.


**Sintaxis**



# Creando `funciones` sin parámetros y sin retornar un valor

```python
print("Matias bienvenido al curso de python")
print("Matias ya estas inscrito en nuestro sistema")
```

```python
# Definicion de la funcion
def saludo():
    print("Matias bienvenido al curso de python")
    print("Matias ya estas inscrito en nuestro sistema")

# invocar funcion

saludo()
```

```python
saludo()
```

# Creando `funciones` con parámetros y sin retornar un valor

```python
# Definicion de la funcion
def saludo(nombre,apellido):
    print(f'{nombre.title()} {apellido.title()} bienvenido al curso de python')
    print(f'{nombre.title()} ya estas inscrito en nuestro sistema')

# invocar funcion

saludo("matias","aguayo")
```

```python
saludo("wilson","garcia")
```

```python
saludo("Goku","son")
```

# Creando `funciones` con parámetros y retornando un valor (`return`)

```python
# Definicion de la funcion
def solicitar_licencia(usuario):
    # Bucle de valicadion de edad
    while True:
        edad = int(input("Ingresa tu edad para validar si eres apto para solicitar licencia de conducir"))

        if edad < 11 or edad > 80:
            print(f'Ingresa una edad valida entre 11 y 80 años. Edad ingresada: {edad}')
        else:
            print(f'Edad ingresada: {edad}')
            print(f'Usuario:{usuario}')
            break # Se detiene cuando la edad ingresada es correcta

    # Evaluacion de condicion para licencia de conducir
    if edad >= 18:
        print(f'El usuario esta habilitado para solicitar licencia de conducir')
        status = True
    else:
        print(f'El usuario no esta habilitado para soliciatar licencia de conducir')
        status = False

    return status
```

```python
# Invocar funcion

respuesta_funcion = solicitar_licencia("Sara")

print(f'Status para solicitar la licencia de conducir: {respuesta_funcion}')
```

```python
respuesta_funcion = solicitar_licencia("Tanjiro")

print(f'Status para solicitar la licencia de conducir: {respuesta_funcion}')
```

```python
saldo = 30000

def cajero_automatico(usuario):
    while True:
        monto = int(input("Ingrese el monto a extraer: "))

        if monto < 5000:
            print(f'El monto minimo a extrer es $5000. Ingresaste: ${monto}\n')

        elif monto > saldo:
            print(f' Monto supera el saldo. Saldo disponible : ${saldo}')

        else:
            print(f'Saldo a sacar ${monto}')
            print(f'Nombre de usuario de cuenta: {usuario}')
            break

    saldo_restante = saldo - monto
    print(f'El usuario extrajo el monto ${monto}. Saldo restante ${saldo_restante}')
    estado_extraccion = True

    return estado_extraccion
    
```

```python
extraccion_dinero = cajero_automatico('James')

print(f'Estado de extraccion de dinero de cajero: {extraccion_dinero}')
```

```python
saldo_tarjeta = 15000

def cargar_tarjeta(usuario,saldo_actual):

    while True:
        monto_cargar = int(input("Digite el monto a cargar: "))

        if monto_cargar < 1000:
            print(f'Monto a cargar debe superar el minimo. Monto ingresado {monto_cargar}')

        elif monto_cargar > 20000:
            print(f'Supera el monto maximo permitido. Monto ingresado {monto_cargar}' )

        elif saldo_actual + monto_cargar > 50000:
            print(f'Tope de saldo superado en la tarjeta')

        else:
            nuevo_saldo = saldo_actual + monto_cargar
            print(f'Carga exitosa para {usuario}. Nuevo saldo es: ${nuevo_saldo}')
            break

    return True
    
```

```python
carga_dinero = cargar_tarjeta("James",saldo_tarjeta)

print(f'Estado de la operacion: {carga_dinero}')
```

```python
saldo_actual = 25000

def trasnferir_dinero(usuario,saldo_disponible):
    while True:
        monto_transferir = int(input("Monto a transferir: "))

        if monto_transferir < 1000:
            print(f'Monto a cargar debe ser desde $1000. monto ingresado {monto_transferir}')

        elif monto_transferir > 500000:
            print(f'Monto execede el maximo. Monto ingresado {monto_transferir}')


        elif monto_transferir > saldo_disponible:
            print(f'Saldo insuficiente. Tu saldo es {saldo_disponible}')

        else:
            nuevo_monto = saldo_disponible - monto_transferir
            print(f'Monto ingresado transferido. {usuario} su nuevo saldo es ${nuevo_monto}')
            break

    return nuevo_monto


```

```python
nuevo_saldo = trasnferir_dinero("Matias",saldo_actual)

print(f'Tu nuevo saldo es ${nuevo_saldo}')

```

```python
mis_puntos = 12000

def canjear_puntos(cliente,puntos_disponibles):
    while True:
        puntos_canjear = int(input("Puntos a canjear: "))

        if puntos_canjear < 500:
            print(f'Los puntos minimos a canjear son desde 500 puntos. Puntos ingresados {puntos_canjear}')

        elif puntos_canjear > 10000:
            print(f'Maximo de puntos canjeados superado , solo se permite 10000 puntos. Puntos ingresados {puntos_canjear}')

        elif puntos_canjear > puntos_disponibles:
            print(f'Los puntos a canjear son mayores a los tuyos. Puntos Ingresados: {puntos_canjear}')

        else:
            nuevos_puntos = puntos_disponibles - puntos_canjear
            print(f'Puntos canjeados {puntos_canjear} y nueva cantidad de puntos {nuevos_puntos}')
            break

    return nuevos_puntos
```

```python
puntos_canjeados = canjear_puntos("Juan",mis_puntos)

print(f'Puntos restantes despues del canje: {puntos_canjeados}')
```
