# `docstrings`

Son cadenas de texto (`str`) que se utilizan para <u>documentar</u> el propósito y funcionamiento de una función. Se encuentran justo después de la definición de la función, clase o módulo, y sirven como documentación interna que describe qué hace el código, cómo se usa, qué parámetros espera y qué valores devuelve.

## Características

* Los docstrings se colocan inmediatamente después de la declaración de una función, entre comillas triples (`"""` o `'''`).

* Puedes acceder a un docstring en tiempo de ejecución utilizando el atributo especial `.__doc__`.

```python
personajes = []

def agregar_personaje(nombre_usuario):
    """_sumary_
    
    Esta función agrega a una lista el personaje suministrado por el usuario

    Args:
        nombre_usuario (str): Nombre del usuario para personalizar

    Returns:
        None.
    """

    while True:
        personaje = input(f"{nombre_usuario.title()}, ingresa el nombre del personaje que quieres agregar: ").title()

        if personaje.isdigit() or personaje.isnumeric() or personaje.isdecimal():
            print(f"Valor no valido {personaje}. {nombre_usuario.title()}, intenta de nuevo...")

        elif (not personaje.isdigit() or not personaje.isnumeric() or not personaje.isdecimal()) and personaje != "Salir" and personaje != "":
            print(f"{nombre_usuario.title()}, has agregado al personaje '{personaje}'")
            personajes.append(personaje)
            print(f"La lista de personajes: {', '.join(personajes)}\n")

        elif personaje == "Salir":
            print(f"saliendo del programa... Adiós {nombre_usuario} 👍🏼")
            print(f"Lista final de personajes: {', '.join(personajes)}")
            break
    
```

```python
nombre_usuario = input("Ingresa ru nombre: ").title()

agregar_personaje(nombre_usuario)
```

```python
print(agregar_personaje.__doc__)
```
