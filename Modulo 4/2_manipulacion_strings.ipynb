# Métodos para manipular un `str`

## `split()`

Divide un `str`, convirténdola en una `lista`.

Opcionallmente se puede indicar el caracter por el cual se quiere hacer la separacion. Si no se indica este caracter de separación se hará con el `espacio en blanco`.

```python
texto = "Hola mundo python"
print(texto.split())
```

```python
frutas = "fresa, mango, maracúya, sandía"
print(frutas.split())
print(type(frutas.split(sep=",")))
```


## `strip()`

Elimina los espacios en blanco (o caracteres especificos) al principio y al final de la cadena.

```python
texto = "     Hola mundo     "
print(texto)
print(texto.strip())
```

```python
texto = "///////Bienvenidos al curso de python//////////"
print(texto)
print(texto.strip("/"))
```

## `lstrip()`

Elimina los espacios (o caracteres) al principio de la cadena.

```python
texto = "///////Bienvenidos al curso de python//////////"
print(texto)
print(texto.lstrip("/"))
```

## `rstrip()`

Elimina los espacios (o caracteres) al final de la cadena.

```python
texto = "///////Bienvenidos al curso de python//////////"
print(texto)
print(texto.rstrip("/"))
```

## `Capitalize()`

Convierte el primer carácter de la cadena a mayúsculas y el resto a minúsculas.

```python
nombre = "goku"
print(nombre.capitalize())
```

## `lower()`

Convierte todos los caracteres de una cadena a minúsculas.

```python
nombre = "GOKU"
print(nombre.lower())
```

## `upper()`

```python
texto = "bienvenidos al curso de python"
print(texto.upper())
```

## `title`

```python
nombre_completo =  "son goku"
print(nombre_completo.title())
```

## `replace()`

Reemplaza todas las ocurrencias de una subcadena old por otra subcadena new.

```python
texto = "Hola mundo"
print(texto.replace("mundo", "Python"))
```
