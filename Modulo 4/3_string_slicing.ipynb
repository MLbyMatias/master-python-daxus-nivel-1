# `Slicing` (segmentación)

El `slicing` (o segmentación) de strings en Python es una técnica que permite extraer una porción o subsecuencia de una cadena de texto (string) utilizando una notación especial.

Es una forma muy poderosa y flexible de trabajar con cadenas, y se puede aplicar no solo a strings, sino también a tipo de secuencias de datos.

# Accediendo a un índice

```python
frase = "Mi nombre es naruto y estoy aprendiendo python"
print(frase[0])
print(frase[1])
print(frase[11])
```

# Accediendo a un `sub-str`

```python
print(frase[0:9])
```

```python
print(frase[0:19])
```

```python
print(frase[-6:])
```

# paso

```python
print(frase[0::2])
print(frase)
```

```python
frase2 = 'Estudiar Python'
print(frase2[0::2]) # Inicia en la primera y salta la otra y asi sucesivamente
```

# Invirtiendo un `str`

```python
frase = "Mi nombre es naruto y estoy aprendiendo python"
print(frase[::-1])
```
