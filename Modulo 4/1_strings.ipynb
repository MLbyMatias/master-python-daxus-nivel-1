# Strings

El tipo de dato string `(str)` es utilizado principalmente para almacenar **texto** (el nombre de una persona, ciudad de origen, marca de ropa, una dirección, etc.). Este *texto* lo podemos entender como una serie de *caracteres* que conforman un `str`. Para formar un `str` en Python lo podemos hacer con comillas dobles (`" "`) o sencillas (`' '`).

## Los `str` son *case sensitive*

```python
"Matias" == "Matías"
```

```python
"Python" == "Python"
```

# Concatenar `str` con el operador aritmético

```python
nombre = "Son"
apellido = "Goku"

print(nombre, apellido)
```

```python
print(nombre + " " + apellido)
```

```python
type("a")
```

```python
print("a" + "b" + "c" + "d")
```

# Los `str` son objetos iterables

```python
print(apellido)
print(apellido[0])
print(apellido[1])
print(apellido[2])
print(apellido[3])
```

```python
lenguaje = "Mi nombre es Matias y estoy estudiando Python"
print(lenguaje[-6])
print(lenguaje[-5])
print(lenguaje[-4])
print(lenguaje[-3])
print(lenguaje[-2])
print(lenguaje[-1])
```
