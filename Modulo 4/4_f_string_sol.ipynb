# f-strings

También llamadas "string literale formateada" (***formated string literal***), las cuales son una forma de formatear cadenas de texto de manera más sencilla y eficiente. Se introdujo en `Python 3.6` y se usa para insertar expresiones dentro de una cadena de texto de forma más directa.

Para utilizar un `f-string`, se coloca una letra `f` o `F` antes de las comillas que definen la cadena, y dentro de la cadena, se pueden poner expresiones o variables entre llaves `{}`.

```python
nombre = "Pepito"
edad = 25

print(f'Hola mi nombre es {nombre} y tengo {edad} años')

```

```python
num_1 = int(input("Ingrese el primer numero: "))
num_2 = int(input("Ingrese el segundo numero: "))

print(f'La suma de los números {num_1} y {num_2} es igual {num_1 + num_2}')
```

# Método `format`

```python
nombre = 'Veggeta'
edad = 27

print("Hola soy {} y tengo {} años".format(nombre,edad))
```

# Método `%`

```python
print('Hola mi nombre es %s y tengo %d años'%(nombre,edad))

# %s = string
# %d = digito
```
