# Diccionarios `dict{key : value}`

Son estructuras de datos que almacenan pares de `clave-valor`. Son una de las estructuras más **versátiles** y útiles en Python debido a su capacidad para **acceder a los datos de manera rápida** mediante claves.

A diferencia de las listas, que utilizan índices numéricos, los diccionarios usan claves (que pueden ser de cualquier tipo inmutable, como cadenas, números o tuplas) para acceder a los valores correspondientes.

## Principales características
* ****Desordenados:**** En versiones anteriores de Python, los diccionarios no mantenían un orden. Sin embargo, a partir de Python 3.7, el orden de los elementos en un diccionario se conserva, es decir, los elementos se almacenan en el mismo orden en que fueron añadidos.

* **Acceso rápido:** El acceso a los valores en un diccionario es muy rápido, ya que internamente utilizan una estructura de datos llamada tabla hash para almacenar las claves y sus valores asociados.

* **Claves únicas:** Las claves en un diccionario deben ser únicas. Si intentas asignar un valor a una clave que ya existe, el valor anterior se sobrescribirá.

* **Valores mutables:** Los valores de un diccionario pueden ser de cualquier tipo de datos, incluidos otros diccionarios, listas, enteros, cadenas, etc. Esto hace que los diccionarios sean muy flexibles.

## Sintáxis

`{"clave" : "valor"}`

# Creando un `dict`

```python
mi_dict = {
    "Nombre": "Goku",
    "Apellido":"Son",
    "Raza":"Sayayin",
    "Planeta":"Veggita"
}

print(mi_dict)
```

# Accediendo a `claves`, `valores` e `items`

```python
mi_dict = {
    "Nombre": "Goku",
    "Apellido":"Son",
    "Edad":25,
    "Raza":"Sayayin",
    "Planeta":"Veggita"
 
}

mi_dict
```

## Método `keys()`

Devuelve una vista de todas las claves del diccionario.

```python
print(mi_dict.keys())
print(type(mi_dict))
```

## Método `values()`

Devuelve una vista de todos los valores del diccionario.

```python
print(mi_dict.values())
print(type(mi_dict))
```

## Método `items()`

Devuelve una vista de todos los pares clave-valor en el diccionario.

```python
print(mi_dict.items())
print(type(mi_dict))
```

# Accediendo a los elementos de un `dict`

## Llamando la `clave`/`llave`

```python
datos_personales = {
    "Nombre": "Goku",
    "Apellido":"Son",
    "Edad":25,
    "Raza":"Sayayin",
    "Planeta":"Veggita"
}

datos_personales
```

```python
print(datos_personales["Raza"])
print(datos_personales["Apellido"])

```

## Método `get()`

Devuelve el valor de una clave específica. Si la clave no existe, devuelve None o un valor predeterminado que tú especifiques.

```python
print(f'Planeta: {datos_personales.get("Planeta")}')
print(f'Nombre: {datos_personales.get("Nombre")}')
print(f'Edad: {datos_personales.get("Edad")}')
```

# Insertando elementos en un `dict`

```python
datos_personales["Hijos"] = ["Gohan","Gotten"]
```

```python
datos_personales
```

```python
print(datos_personales["Hijos"][0])
print(datos_personales["Hijos"][1])
```

# Actualizando elementos en un `dict`

```python
datos_personales["Edad"] = 26
datos_personales['Nombre'] = 'Goku Rafael'

datos_personales
```

# Eliminando elementos de un `dict`

## Método `pop()`

Elimina y devuelve el valor asociado a la clave dada. Si la clave no existe y no se proporciona un valor default, lanza una excepción/error KeyError.

```python
datos_personales.pop("Planeta")
```

```python
datos_personales
```

## Método `popitem()`

Elimina y devuelve el último par clave-valor del diccionario (en versiones anteriores de Python, eliminaba un par aleatorio).

```python
hijos = datos_personales.popitem()

print(hijos)

datos_personales
```

## Método `clear()`

Elimina todos los elementos del diccionario.

```python
datos_personales.clear()

datos_personales
```
