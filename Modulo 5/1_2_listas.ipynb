# Listas - `list[]`

Una lista es una colección ***ordenada*** y ***mutable*** de elementos. Es una de las estructuras de datos más flexibles y usadas en Python.

Puedes almacenar elementos de ***diferentes tipos*** dentro de una misma lista, como enteros, cadenas de texto, otras listas, etc. Las listas se crean utilizando corchetes `[]` y los elementos están separados por comas.

## Principales características

* **Ordenadas:** los elementos tienen un **índice**, lo que significa que su posición en la lista es importante y se puede acceder a ellos usando ese índice.
* **Mutables:** es posible **modificar** el contenido de una lista (agregar, eliminar, cambiar elementos) después de haberla creado.
* **Permiten duplicados:** pueden existir elementos repetidos dentro de una lista.

## Creando una lista

```python
lista_dbz = ['Goku','Gohan','Veggeta']
print(type(lista_dbz))
print(lista_dbz)
```

## Creando una lista con `diferentes` tipos de datos / objetos

```python
lista_heterogenea = [25,25,True,"Hola",1.80,"Python"]
print(type(lista_heterogenea))
print(lista_heterogenea)

```

```python
lista_anidada = [2000,"Matias",[1,23,24,True]]
print(type(lista_anidada))
print(lista_anidada)
```

## Accediendo a un elemento de una lista por medio del `índice`

```python
lista_dbz = ["Goku","Gohan","Vegetta","Trunks"]

print(lista_dbz[0])
print(lista_dbz[1])
print(lista_dbz[2])
print(lista_dbz[-1])
```

```python
print(lista_anidada[1])
print(lista_anidada[2])
print(lista_anidada[2][-1])
```

## Utilizando `slicing` en una lista

```python
personajes = ["Naruto","Luffy","Pikachu","Bart","Homero","Usopp"]

print(personajes)
print(47*"-")
print(personajes[0:2])
print(personajes[-3:])
print(personajes[2:])
print(personajes[2:4])
```

## Operadores `in` y `not in`

```python
print(personajes)

print("Gohan" in personajes)
print("Bart" in personajes)
print("Homero" not in personajes)
```

## `Manipulando` elementos de una lista

### Método `append()`

Añade un elemento al **final** de la lista.

```python
print(personajes)
personajes.append("Maggie")
print(personajes)
```

### Método `insert()`

Inserta un elemento en una posición específica de la lista.

```python
print(f'Personajes antes de aplicar el método instert: {personajes}')
```

```python
personajes.insert(2,"Itadori")
print(f'Personajes despues de aplicar el método instert: {personajes}')
```

### Método `count()`

Cuenta cuántas veces aparece un valor en la lista.

```python
lista_numeros = [12,123,123123,123,123,123,12,312,3]
print(lista_numeros)
```

```python
print(lista_numeros.count(123))
```

### Método `reverse()`

Invierte el orden de los elementos en la lista.

```python
mi_lista = [1,2,3]
mi_lista.reverse()

print(mi_lista)
```

### Método `sort()`

Ordena la lista en orden ascendente (**por defecto**). Si le pasas reverse=True, la ordena en orden descendente.

```python

mi_lista = [3,2,1]

mi_lista.sort()

print(mi_lista)

mi_lista.sort(reverse=True)

print(mi_lista)
```

### Método `pop()`

Elimina y devuelve el elemento en una posición específica (por defecto elimina el último elemento).

```python
print(personajes)

print(personajes.pop())

print(personajes)
```

```python
print(personajes)

print(personajes.pop(-2))

print(personajes)
```

### Método `remove()`

Elimina la primera aparición de un valor en la lista.

```python
lista_repetidos = [1,2,3,4,4,5]

print(lista_repetidos)
```

```python
lista_repetidos.remove(4)

print(lista_repetidos)
```

### Método `clear()`

Elimina todos los elementos de la lista, dejándola vacía.

```python
print(lista_repetidos)
```

```python
lista_repetidos.clear()
print(lista_repetidos)
```

### Comando `del`

Podemos excluir un elemento de una lista utilizando el comando `del`, seguido de la lista y el índice del elemento que será excluido.

```python
lista_lenguajes = ["Python","Java","C++"]

print(lista_lenguajes)
```

```python
del lista_lenguajes[1]
print(lista_lenguajes)
```

## `Mutando` elementos de una lista

```python
personajes = ['Luffy', 'Goku', 'Pikachu', 'Bart', 'Usopp']

print(personajes)
```

```python
personajes[2] = "Zoro"
```

```python
print(personajes)
```
