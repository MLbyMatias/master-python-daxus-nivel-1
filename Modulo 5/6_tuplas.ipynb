# Tuplas - `tuple()`

Estructuras de datos muy similares a las listas, pero con una **diferencia importante**: son **<u>inmutables</u>**, lo que significa que no puedes cambiar sus elementos una vez que la tupla ha sido creada. Las tuplas son muy útiles cuando necesitas asegurarte de que los datos no sean modificados, o cuando quieres un contenedor más eficiente en términos de memoria.

## Principales características

* Se definen con paréntesis `()`.

* Son **<u>inmutables</u>** (seguridad) (no puedes cambiar, agregar ni eliminar elementos).

* Son **ordenadas**, lo que significa que puedes acceder a los elementos por su índice.

* Pueden contener **cualquier tipo de dato** (números, cadenas, listas, etc.).

* Debido a que son inmutables, las tuplas suelen ser más **rápidas** que las listas en operaciones de acceso.

# Desempaquetado de una `tuple`

Acción de extraer los elementos de una tupla y asignarlos a variables de manera simultánea. Es una forma concisa y eficiente de acceder a los elementos de una tupla (o cualquier iterable) y asignarlos a variables individuales en una sola línea.

## Desempaquetado básico

```python
simpsons = ("Homero","Marge","Bart","Lisa")

padre, madre, hijo , hija = simpsons

print(padre)
print(madre)
print(hijo)
print(hija)
```

## Desempaquetado con tuplas de longitud variable

```python
print(simpsons)
padre, * resto_familia = simpsons
print(type(padre))
print()
print(resto_familia)
print(type(resto_familia))
```

# Desempaquetado ignorando elementos

Si solo te interesa algunos de los elementos de la tupla, puedes ignorar los elementos que no te interesan utilizando el símbolo de subrayado (`_`), que es una convención común en Python.

```python
tupla = (1,2,3,4)

a,_,c,_ = tupla

print(a)
print(c)
print("------------------")
print(_)
```
