# Listas - `list[]`

Una lista es una colección ***ordenada*** y ***mutable*** de elementos. Es una de las estructuras de datos más flexibles y usadas en Python.

Puedes almacenar elementos de ***diferentes tipos*** dentro de una misma lista, como enteros, cadenas de texto, otras listas, etc. Las listas se crean utilizando corchetes `[]` y los elementos están separados por comas.

## Principales características

* **Ordenadas:** los elementos tienen un **índice**, lo que significa que su posición en la lista es importante y se puede acceder a ellos usando ese índice.
* **Mutables:** es posible **modificar** el contenido de una lista (agregar, eliminar, cambiar elementos) después de haberla creado.
* **Permiten duplicados:** pueden existir elementos repetidos dentro de una lista.

# Concatenación (`+`) y repetición (`*`) de listas

```python
lista_1 = [1,2,3,4]
lista_2 = [5,6,7,8]

lista = lista_1 + lista_2

print(lista)
```

```python
lista = ['Homero','Marge','Lisa','Bart']

lista_repeticion = lista * 3

print(lista_repeticion)
```

# Método `len()`

Indica el número de elementos que contiene un iterable (en este caso `list`)

```python

print(lista_repeticion)
lista_repeticion_len = len(lista_repeticion)
print(f'La cantidad de elementos de esta lista es: {lista_repeticion_len}')
print(lista)
lista_len = len(lista)
print(f'La cantidad de elementos de esta lista es: {lista_len}')

```

# Método `index()`

Se utiliza para encontrar el **índice** de la <u>primera aparición</u> de un elemento en una lista. Si el elemento no está en la lista, Python lanzará un error llamado `ValueError`.

```python
print(lista)
print(lista.index("Bart"))
```

```python
print(lista.index("Lisa"))
```

```python
print(lista.index("Barie"))
```

# Método `sorted()`

**Distinción importante:** `sorted()` no es un método de las listas, sino una función `built-in` que devuelve una nueva lista ordenada, dejando la lista original sin cambios.

```python
lista_desordenada = [1, 5, 3, 2, 4]
print(lista_desordenada)
lista_ordenada = sorted(lista_desordenada)
print(lista_ordenada)
```

# Método alternativo con índices para `reverse()`

```python
lista = [1,2,3,4,5,6]
print(lista[::-1])

lista.reverse()
print(lista)
```

# Método `copy()`

se utiliza para crear una copia superficial (shallow copy) de una lista. Esto significa que el nuevo objeto creado será una nueva lista, pero los elementos dentro de la lista no son copiados de manera profunda. En otras palabras, si los elementos dentro de la lista son <u>objetos mutables</u> (como otras listas), se mantienen como referencias a los mismos objetos. Si los elementos son inmutables (como números o cadenas de texto), se copian como valores.

```python
lista_original = [1,2,3,4,5]
copia = lista_original.copy()

print(copia)
```

```python
copia.append(5)

print("Lista original ", lista_original)
print("Copia: ", copia)
```

# Método `deepcopy()`

El método `deepcopy()` en Python es parte del módulo copy, y se utiliza para crear una **<u>copia completa</u>** de un objeto, lo que significa que crea una nueva instancia del objeto y de todos los objetos anidados dentro de él, es decir, que las copias sean <u>completamente independientes</u>. Esto es útil cuando necesitas evitar efectos secundarios inesperados en los objetos originales al modificar la copia.

```python
import copy

lista_original = [1,[2,3],4]
copia = copy.deepcopy(lista_original)

copia[1].append(5)


print("Lista original ", lista_original)
print("Copia: ", copia)
```

**En resumen**

* `copy()`: Crea una copia superficial. Solo la estructura principal (como la lista) es copiada, pero los objetos internos siguen siendo las mismas referencias en memoria.

* `deepcopy()`: Crea una copia completamente independiente, copiando todos los objetos anidados dentro de la estructura.
