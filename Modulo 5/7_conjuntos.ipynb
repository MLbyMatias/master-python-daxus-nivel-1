# Conjuntos - `set()`

Son una estructura de datos que almacena elementos **<u>únicos</u>** sin un orden específico. Los conjuntos se definen usando llaves `{}` o con la función `set()`. A diferencia de las listas y tuplas, los conjuntos **<u>no permiten elementos duplicados</u>**.

## Principales características

* Son **no ordenados**, lo que significa que <u>no</u> puedes acceder a los elementos por <u>índice</u>.

* Son **mutables**, pero <u>no</u> permiten elementos <u>duplicados</u>.

* Son **iterables**, lo que significa que puedes <u>recorrer</u> estos objetos.

# Creando un `set`

```python
conjunto = {1,2,3,4,5}

print(conjunto)
print(type(conjunto))
```

```python
conjunto_repetido = {1,2,2,3,4,5}

print(conjunto_repetido)
```

# Manipulando elementos de un `set`

## Método `add()`

Agrega el elemento x al conjunto. Si el elemento ya existe, no hace nada (no permite duplicados).

```python
print(conjunto)
```

```python
conjunto.add(6)

print(conjunto)
```

```python
conjunto.add("Guts")

print(conjunto)
```

```python
conjunto.add(False)

print(conjunto)
```

```python
conjunto.add(1.234)
print(conjunto)
```

```python
conjunto.add(("Naruto","Usumaki"))
print(conjunto)
```

```python
conjunto.add([7,8,9])
print(conjunto)
```

```python
conjunto[3]
```

# Eliminando elementos de un `set`

## Método `remove()`

Elimina el elemento x del conjunto. Si el elemento no está presente, lanza una excepción KeyError.

```python
mi_conjunto = {1,2,3,4}
print(mi_conjunto)

mi_conjunto.remove(3)
print(mi_conjunto)
```

## Método `pop()`

Elimina y devuelve un elemento **<u>aleatorio</u>** del conjunto. Si el conjunto está vacío, lanza una excepción KeyError.

```python
mi_conjunto = {1,2,3,4}
print(mi_conjunto)

elemento = mi_conjunto.pop()
print(elemento)

print(mi_conjunto)
```

## Método `clear()`

Elimina todos los elementos del conjunto.

```python
print(mi_conjunto)
mi_conjunto.clear()

print(mi_conjunto)
```

# Operaciones con `sets`

## Método `union()`

Devuelve un conjunto con los elementos combinados de dos conjuntos (**sin duplicados**).



```python
conjunto_1 = {1,2,3}
conjunto_2 = {4,2,5}

print(conjunto_1.union(conjunto_2))
print(conjunto_1 | conjunto_2)
```

# Método `intersection()`

Devuelve un conjunto con los elementos **comunes** entre dos conjuntos.



```python
conjunto_1 = {1,2,3}
conjunto_2 = {2,3,5}
print(conjunto_1.intersection(conjunto_2))
print(conjunto_1 & conjunto_2)
```
