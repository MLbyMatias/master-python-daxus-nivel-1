# Comandos `in` e `del`

### Aplica `in`

Método directo para validar la existencia de un elemento en estructuras de datos. Especialmente en `sets` y `dicts` ya que utilizan ***tablas hash*** en su implementación en Python.

* `list`

* `dict`: validación por ***keys*** y ***values***.

* `tuple`

* `set`

### Aplica `del`

Útil para prescindir de variables que almacenan grandes cantidades de datos cuando ya no son necesarias.

* `list`: eliminación por ***index*** y lista completa.

* `dict`: eliminación por ***key*** y diccionario completo.

* `tuple`: eliminación completa únicamente (estructura de datos ***inmutable***).

* `set`: eliminación completa únicamente (estructura de datos ***desordenada***).

# `list`

## `in`

```python
personaje = ["Naruto", "Goku", "Poikachu", "Luffy", "Bart"]
```

```python
"Naruto" in personaje
```

```python
"Veggeta" in personaje
```

## `del`

```python
print(personaje)
```

```python
del personaje[-2]
```

```python
print(personaje)
```

```python
del personaje
print(personaje)
```

# `dict`

## `in`

```python
datos_personales = {
    "Nombre" : "Goku",
    "Apellido" : "Son",
    "Edad" : 25,
    "Raza" : "Sayayin",
    "Planeta" : "Veggita"
}
```

```python
"Edad" in datos_personales.keys()
```

```python
"Daima" in datos_personales.values()
```

```python
"Veggita" in datos_personales.values()
```

## `del`

```python
del datos_personales["Planeta"]
print(datos_personales)
```

```python
del datos_personales
print(datos_personales)
```

# `tuple`

## `in`

```python
usuario = ("Juancho", "Hernandez", True)
```

```python
"Hernandez" in usuario
```

```python
"Fernandez" in usuario
```

## `del`

```python
del usuario
print(usuario)
```

# `set`

## `in`

```python
mi_conjunto = {1,2,3,4}
```

```python
3 in mi_conjunto
```

```python
10 in mi_conjunto
```

## `del`

```python
del mi_conjunto
print(mi_conjunto)
```
