# Conjuntos - `set()`

Son una estructura de datos que almacena elementos **<u>únicos</u>** sin un orden específico. Los conjuntos se definen usando llaves `{}` o con la función `set()`. A diferencia de las listas y tuplas, los conjuntos **<u>no permiten elementos duplicados</u>**.

## Principales características

* Son **no ordenados**, lo que significa que <u>no</u> puedes acceder a los elementos por <u>índice</u>.

* Son **mutables**, pero <u>no</u> permiten elementos <u>duplicados</u>.

* Son **iterables**, lo que significa que puedes <u>recorrer</u> estos objetos.

# Transformando un `set` a `list`

```python
conjunto = {1,2,3,4,5}

print(conjunto)
print(type(conjunto))
```

```python
conjunto_a_lista = list(conjunto)
print(conjunto_a_lista)
print(type(conjunto_a_lista))
```

# Método `discard()`

Este método elimina el elemento especificado del conjunto. Si el elemento no está presente, no hace nada y no genera error.

```python
conjunto = {1,2,3,False,"Python"}
print(conjunto)
```

```python
conjunto.discard(3)
print(conjunto)
```

```python
conjunto.discard("Java")
print(conjunto)
```

```python
conjunto.discard(False)
print(conjunto)
```

# Método `difference()`

Este método devuelve un nuevo conjunto con los elementos que están en el conjunto original pero no en los otros conjuntos que se pasen como argumento. No modifica el conjunto original.

```python
conjunto_1 = {1,2,3,4}
conjunto_2 = {3,4,5,6}

diferencia = conjunto_1.difference(conjunto_2)
print(diferencia)
```

# Método `issubset()`

Este método verifica si el conjunto actual es un subconjunto del conjunto proporcionado. Devuelve `True` si todos los elementos del conjunto actual están en el conjunto pasado como argumento.

```python
conjunto_1 = {1,2}
conjunto_2 = {1,2,3,4}
print(conjunto_1.issubset(conjunto_2))
```

```python
conjunto_3 = {5,6}
print(conjunto_3.issubset(conjunto_2))
```

# Método `issuperset()`

Este método verifica si el conjunto actual es un superconjunto del conjunto proporcionado. Devuelve True si todos los elementos del conjunto pasado como argumento están en el conjunto actual.

```python
conjunto_1 = {1,2,3,4}
conjunto_2 = {1,2}

print(conjunto_1.issuperset(conjunto_2))
```

```python
conjunto_3 = {5,6}
print(conjunto_1.issuperset(conjunto_3))
```
