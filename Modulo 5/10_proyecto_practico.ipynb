# 🛒 Sistema de control de inventario 

### 🏗️ Funcionalidades

✅ `Listas` para almacenar múltiples productos sin usar estructuras de repetición o control

✅ `Diccionarios` para estructurar los datos (`nombre`, `cantidad`, `precio_unitario`, `valor_total`)

✅ Manipulación de cadenas (`strip()`, `title()`) para estandarizar la entrada del usuario

✅ Cálculo automático del valor total del inventario

✅ Separación visual con caracteres(`"-" * 40`) para un diseño más organizado

```python
print("🛒 Bienvenidos")

# Definir inventario vacio
inventario = []
IVA = 0.19



# Solicitando información al usuario
producto1 = input("Ingrese el nombre del producto: ").title().strip()
cantidad1= int(input("Ingrese la cantidad: "))
precio_unitario1 = float(input("Precio del producto: "))

# Calculo del valor total del inventario
valor_total_1 = cantidad1 * precio_unitario1

producto_1 = {
    "nombre":producto1,
    "cantidad":cantidad1,
    "precio":precio_unitario1,
    "valor_total":valor_total_1
}

inventario.append(producto_1)


# Mostrando los datos formateados
print("\n📦 Inventario Inicial")
print(f"Producto:{inventario[0]["nombre"]}",)
print(f"Cantidad:{inventario[0]["cantidad"]}",)
print(f"Precio unitario: ${inventario[0]['precio']}", )
print(f"Valor total del inventario: ${inventario[0]['valor_total']:.2f}", )
print("🛒 Sistema de Control de Inventario")

# Solicitando información al usuario
producto2 = input("Ingrese el nombre del producto: ").title().strip()
cantidad2 = int(input("Ingrese la cantidad: "))
precio_unitario2 = float(input("Precio del producto: "))

# Cálculo del valor total del inventario
valor_total_2 = cantidad2 * precio_unitario2

producto_2 = {
    "nombre":producto2,
    "cantidad":cantidad2,
    "precio":precio_unitario2,
    "valor_total":valor_total_2
}

inventario.append(producto_2)

iva_producto1 = valor_total_1 * IVA
iva_producto2 = valor_total_2 * IVA

total_iva = iva_producto1 + iva_producto2

valor_total = valor_total_1 + valor_total_2

total_con_iva = valor_total + total_iva



# Mostrando los datos formateados
print("\n📦 Inventario Inicial")
print(f"Producto:{inventario[1]["nombre"]}",)
print(f"Cantidad:{inventario[1]["cantidad"]}",)
print(f"Precio unitario: ${inventario[1]["precio"]:.2f}", )
print(f"Valor total del inventario: ${inventario[1]["valor_total"]:.2f}", )
print("🛒 Sistema de Control de Inventario")



print("-" * 40)
print(f'El valor total es de: ${valor_total}')
print(f'El IVA total de esta boleta es de: {total_iva}')
print(f'El valor con IVA total es de: {total_con_iva}')





# inventario[0] Entras al primer elemento de la lista (que es el primer diccionario).
# ["nombre"] Entras a la clave "nombre" dentro de ese diccionario específico para extraer su valor.
```
